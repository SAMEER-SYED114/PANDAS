```python
#joining,merge,concatinate
import numpy as np
import pandas as pd
#merging 2 data frames


employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
pd.merge(employees,salaries,on="employee_id",how='inner')
pd.merge(employees,salaries,on="employee_id",how='outer')
pd.merge(employees,salaries,on='employee_id',how='left',)
pd.merge(employees,salaries,on='employee_id',how='right')

```

```python
#data extraction practice
import pandas as pd 
import numpy as np
df=pd.read_csv(r"C:\Users\syed naziya sultana\Downloads\anime.csv")
df
```

```python
def extract_episodes(txt):
 check=False
 data=""
 for i in txt:
  if i==')':
   check=False
   return data
  if check==True:
   data+=i
  if i=='(':
    check=True




        
```

```python
#extracting episods and converting ino str and separate colum for episodes
df['Episodes']=df["Title"].apply(extract_episodes)
df["Episodes"]=df['Episodes'].str.replace(' eps','')
df["Episodes"].astype(int)
df


```

```python
#extraction of time 
def extract_time(txt):
 data=''
 for i in range(len(txt)):
  if txt[i]==')':
   for j in range(i+1,i+20):
    data+=txt[j]
 return data
   
```

```python
df['Time_year']=df['Title'].apply(extract_time)
df
```

```python
#months calculation
import pandas as pd

# Split into start and end, parse as dates
df[['start', 'end']] = df['Time_year'].str.split(' - ', expand=True)
df['start'] = pd.to_datetime(df['start'], format='%b %Y')
df['end'] = pd.to_datetime(df['end'], format='%b %Y')

# Calculate number of months
df['months'] = (df['end'].dt.year - df['start'].dt.year) * 12 + \
               (df['end'].dt.month - df['start'].dt.month)
df['months_inclusive'] = df['months'] + 1
df
```
